# EGONETS dataset exploration

## Objective
Inspect verified source files and document schema, graph directionality, attributes, class encoding, duplicates, missing values, and per-variant counts.

## Setup
Run `python scripts/verify_dataset.py` and `python scripts/inspect_dataset.py` from the repository root after placing the original files in `data/raw/`. The format-specific package loader remains unimplemented.

## Status
The verified source files are in data/raw/. Use the scripts above for actual schema checks before adding analyses.

1. Source and checksums
2. File inventory and schema
3. Graph and label validation
4. Variant-level descriptive statistics
5. Limitations and unresolved parsing decisions
